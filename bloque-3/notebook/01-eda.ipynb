# Análisis Exploratorio de Datos (EDA)
## Liga de Fútbol - Preparación para Modelado Predictivo

**Objetivo**: Explorar el dataset de partidos de fútbol para preparar nuestro trabajo de modelado predictivo.

**Duración**: ~60 minutos

## 1. Configuración y Carga de Datos

```python
# Importar librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Configuración de visualización
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (12, 6)
pd.set_option('display.max_columns', None)

# Cargar el dataset
data_path = Path('../data/datos_liga_futbol.csv')
df = pd.read_csv(data_path)

print(f"✓ Dataset cargado: {df.shape[0]} partidos, {df.shape[1]} columnas")
```

## 2. Inspección Inicial y Calidad de Datos

```python
# Ver primeras filas
print("PRIMERAS 5 FILAS:")
display(df.head())

# Información del dataset
print("\nINFORMACIÓN DEL DATASET:")
df.info()

# Verificar calidad de datos
print("\nCALIDAD DE DATOS:")
print(f"- Valores nulos: {df.isnull().sum().sum()}")
print(f"- Registros duplicados: {df.duplicated().sum()}")
if df.isnull().sum().sum() == 0 and df.duplicated().sum() == 0:
    print("✓ Dataset limpio y listo para análisis")
```

## 3. Análisis de Resultados (Variable Objetivo)

**¿Qué queremos predecir?** El resultado del partido: Victoria Local, Victoria Visitante, o Empate

```python
# Distribución de resultados
resultados = df['Resultado'].value_counts()
resultados_pct = (resultados / len(df) * 100).round(1)

print("DISTRIBUCIÓN DE RESULTADOS:")
for resultado, count in resultados.items():
    print(f"  {resultado}: {count} partidos ({resultados_pct[resultado]}%)")

# Ventaja de jugar en casa
victorias_local = len(df[df['Resultado'] == 'Victoria Local'])
victorias_visitante = len(df[df['Resultado'] == 'Victoria Visitante'])
ventaja_local = (victorias_local - victorias_visitante) / len(df) * 100
print(f"\n💡 VENTAJA DE JUGAR EN CASA: {ventaja_local:.1f} puntos porcentuales")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico de barras
resultados.plot(kind='bar', ax=ax1, color=['green', 'red', 'orange'])
ax1.set_title('Distribución de Resultados', fontsize=14, fontweight='bold')
ax1.set_xlabel('Tipo de Resultado')
ax1.set_ylabel('Cantidad de Partidos')
ax1.tick_params(axis='x', rotation=45)
ax1.grid(axis='y', alpha=0.3)

# Gráfico de pastel
resultados.plot(kind='pie', ax=ax2, autopct='%1.1f%%', colors=['green', 'red', 'orange'])
ax2.set_title('Proporción de Resultados', fontsize=14, fontweight='bold')
ax2.set_ylabel('')

plt.tight_layout()
plt.show()
```

## 4. Análisis de Goles

Los goles son una característica importante para entender el resultado del partido.

```python
# Estadísticas de goles
print("ESTADÍSTICAS DE GOLES:")
print(f"  Promedio goles local: {df['Goles_Local'].mean():.2f}")
print(f"  Promedio goles visitante: {df['Goles_Visitante'].mean():.2f}")

# Total de goles por partido
df['Goles_Total'] = df['Goles_Local'] + df['Goles_Visitante']
print(f"  Promedio goles por partido: {df['Goles_Total'].mean():.2f}")
print(f"  Partido con más goles: {df['Goles_Total'].max()}")

# Visualización
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Histograma goles local
axes[0].hist(df['Goles_Local'], bins=range(0, df['Goles_Local'].max()+2), 
             color='blue', alpha=0.7, edgecolor='black')
axes[0].set_title('Distribución Goles Local', fontweight='bold')
axes[0].set_xlabel('Goles')
axes[0].set_ylabel('Frecuencia')
axes[0].grid(alpha=0.3)

# Histograma goles visitante
axes[1].hist(df['Goles_Visitante'], bins=range(0, df['Goles_Visitante'].max()+2),
             color='red', alpha=0.7, edgecolor='black')
axes[1].set_title('Distribución Goles Visitante', fontweight='bold')
axes[1].set_xlabel('Goles')
axes[1].set_ylabel('Frecuencia')
axes[1].grid(alpha=0.3)

# Boxplot comparativo
axes[2].boxplot([df['Goles_Local'], df['Goles_Visitante']], 
                labels=['Local', 'Visitante'])
axes[2].set_title('Comparación Local vs Visitante', fontweight='bold')
axes[2].set_ylabel('Goles')
axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Características Clave: Habilidades y Rachas

Estas son las variables predictoras principales que usaremos en nuestros modelos.

```python
# Estadísticas de habilidades
print("HABILIDADES DE LOS EQUIPOS:")
print(f"  Rango de habilidad: {df['Habilidad_Local'].min():.0f} - {df['Habilidad_Local'].max():.0f}")
print(f"  Promedio habilidad local: {df['Habilidad_Local'].mean():.1f}")
print(f"  Promedio habilidad visitante: {df['Habilidad_Visitante'].mean():.1f}")

# Crear feature: Diferencia de habilidad
df['Diferencia_Habilidad'] = df['Habilidad_Local'] - df['Habilidad_Visitante']
print(f"\n💡 DIFERENCIA DE HABILIDAD (Local - Visitante):")
print(f"  Promedio: {df['Diferencia_Habilidad'].mean():.2f}")
print(f"  Rango: {df['Diferencia_Habilidad'].min():.0f} a {df['Diferencia_Habilidad'].max():.0f}")

# Estadísticas de rachas
print(f"\nRACHAS:")
print(f"  Promedio racha local: {df['Racha_Local'].mean():.2f}")
print(f"  Promedio racha visitante: {df['Racha_Visitante'].mean():.2f}")
print(f"  Racha máxima alcanzada: {max(df['Racha_Local'].max(), df['Racha_Visitante'].max())}")

# Visualización
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distribución de diferencia de habilidades
axes[0].hist(df['Diferencia_Habilidad'], bins=30, color='green', alpha=0.7, edgecolor='black')
axes[0].axvline(0, color='red', linestyle='--', linewidth=2, label='Igualdad')
axes[0].set_title('Diferencia de Habilidades (Local - Visitante)', fontweight='bold')
axes[0].set_xlabel('Diferencia de Habilidad')
axes[0].set_ylabel('Frecuencia')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Scatter plot: Diferencia de habilidad vs Resultado
resultado_num = df['Resultado'].map({'Victoria Local': 1, 'Empate': 0, 'Victoria Visitante': -1})
axes[1].scatter(df['Diferencia_Habilidad'], resultado_num, alpha=0.4)
axes[1].axhline(0, color='orange', linestyle='--', linewidth=1, label='Empate')
axes[1].axvline(0, color='red', linestyle='--', linewidth=1, label='Igualdad habilidad')
axes[1].set_title('Diferencia de Habilidad vs Resultado', fontweight='bold')
axes[1].set_xlabel('Diferencia de Habilidad (Local - Visitante)')
axes[1].set_ylabel('Resultado (-1: Visitante, 0: Empate, 1: Local)')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Matriz de Correlación

Identifiquemos las relaciones entre variables para feature engineering.

```python
# Seleccionar variables numéricas relevantes
vars_numericas = ['Goles_Local', 'Goles_Visitante', 'Habilidad_Local', 
                  'Habilidad_Visitante', 'Racha_Local', 'Racha_Visitante',
                  'Diferencia_Habilidad']

correlacion = df[vars_numericas].corr()

print("CORRELACIONES IMPORTANTES:")
print(f"  Habilidad_Local vs Goles_Local: {correlacion.loc['Habilidad_Local', 'Goles_Local']:.3f}")
print(f"  Habilidad_Visitante vs Goles_Visitante: {correlacion.loc['Habilidad_Visitante', 'Goles_Visitante']:.3f}")
print(f"  Diferencia_Habilidad vs Goles_Local: {correlacion.loc['Diferencia_Habilidad', 'Goles_Local']:.3f}")

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlacion, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Matriz de Correlación de Variables', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print("\n💡 INSIGHTS:")
print("  - La diferencia de habilidad es un predictor potencial del resultado")
print("  - Las rachas tienen correlación baja, pero pueden capturar momentum")
print("  - Hay ventaja de jugar en casa (43.2% vs 37.5% victorias)")
```
